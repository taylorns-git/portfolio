```python
import os
import csv

grades = []

def add_grade():
    print('\n= Add New Grade =')

    student = input('Student name: ')
    subject = input('Class subject: ')

    # Loop to ensure the user enters a valid number for the grade.
    while True:
        try:
            grade = float(input('Enter grade: '))

            if grade < 0:
                print('Grade cannot be negative. Please try again.')
            else:
                break
        except ValueError:
            print('Invalid grade. Please enter a numeric value.')

    # Create a dictionary to hold the details of the new grade.
    gradebook = {
        'student': student,
        'subject': subject,
        'grade': grade,
    }

    # Add the newly created grade dictionary to the global grades list.
    grades.append(gradebook)
    print('Grade added successfully!')

def view_grades():
    print('\n= View Grades = ')

    if not grades:
        print('No grades found yet. Add some grades first!')
        return

    for gradebook in grades:
        student = gradebook['student']
        subject = gradebook['subject']
        grade = gradebook['grade']

        print(f'Student {student} | Subject {subject} | Grade {grade:.2f}')

def save_grades():
    if not grades:
        print('No grades to save.')
        return

    print('\nSaving grades to gradebook.csv...')

    with open('gradebook.csv', 'w', newline='') as file:

        fieldnames = ['student', 'subject', 'grade']

        writer = csv.DictWriter(file, fieldnames=fieldnames)

        writer.writeheader()
        writer.writerows(grades)
    print('Grades saved successfully to gradebook.csv!')

def load_grades():
    global grades
    print('\nLoading grades from gradebook.csv.')

    if os.path.exists('gradebook.csv'):
        grades.clear()
        with open('gradebook.csv', 'r', newline='') as file:
            reader = csv.DictReader(file)
            for row in reader:
                try:
                    row['grade'] = float(row['grade'])
                    grades.append(row)
                except ValueError:
                    print(f'Warning: could not convert grade for row: {row}.')
        print(f'Successfully loaded {len(grades)} grades from gradebook.csv.')
    else:
        print('No grades file (gradebook.csv) found, starting with an empty grade.')

def main_menu():
    # Automatically load grades.
    load_grades()


    while True:
        print('\n= Gradebook menu =')
        print('1. Add New Grade')
        print('2. View Grades')
        print('3. Save Grades')
        print('4. Exit')

        choice = input('Enter your choice: ')

        if choice == '1':
            add_grade()
        elif choice == '2':
            view_grades()
        elif choice == '3':
            save_grades()
        elif choice == '4':
            print('Exiting Gradebook. Goodbye!')
            save_grades()
            break
        else:
            print('Invalid choice. Please try again.')

if __name__ == '__main__':
    main_menu()
```
